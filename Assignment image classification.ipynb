## **Q1.** What is a convolutional Neural network (CNN), and how does it differ from traditional fully connected neural networks in terms of architecture and performance on image data?

1. A Convolutional neural nerwork is a
type of deep learning model specially designed to to work with grid like data , such as images.

* An image is basically a grid pixels (height*width*channels like RGB)  
* CNN use Convulational operations to automatically detect important features (edges,textures, shapes,objects)without requiring manual feature engineering.



2. CNN's are made up of layers like:

*   Convolutional layer- Applies filters that slide over the image to extract features (e.g, detecting edges , corners ,patterns).
*   Pooling layer- reduces the spatial size, keeping impoertnant features but reducing computation.
*   Fully connected layer(Dense layer)- works like a traditional neural network at the end,combining .extracted
features for classification or prediction.
   


3. Difference from traditinal fully connected neural networks.

| **Aspect**                | **Fully Connected NN (FCNN)**                                                  | **Convolutional NN (CNN)**                                                             |
| ------------------------- | ------------------------------------------------------------------------------ | -------------------------------------------------------------------------------------- |
| **Architecture**          | Every neuron connected to every other neuron. No spatial structure.            | Uses local connections (filters) that scan over input. Preserves spatial structure.    |
| **Parameters**            | Extremely large for images (e.g., 1M+ weights for a single small image).       | Much fewer parameters due to weight sharing (same filter slides over the whole image). |
| **Feature Extraction**    | Manual feature engineering needed (edges, textures).                           | Learns features automatically at different layers (low → edges, high → objects).       |
| **Performance on Images** | Poor – doesn’t scale well due to too many parameters and no spatial awareness. | Excellent – state-of-the-art in image recognition, object detection, etc.              |
| **Overfitting**           | High risk because of too many parameters.                                      | Lower risk since parameters are reduced with shared filters.                           |


## **Q2**. Discuss the architecture of LeNet-5 and explain how it laid the foundation for modern deep learning models in computer vision. include references to it's  original research paper.

1. What is LeNet-5?
*  LeNet-5 is one of the earliest convolutional neural network (CNN) architectures, proposed by **Yann LeCun et al**. in 1998 in the paper.
it was primarily designed for handwritten digit recongnition (MNIST dataset,checks,zip codes,etc.).
*  LeNet-5 became the blueprint for modern cnn's because it showed how convolution,pooling and fully connected layers can be combined into an end to end trainable system.


2. LeNet-5 Architecture

LeNet-5 consists of 7 layers (excluding input), with learnable parameters in convolutional and fully connected layers.

Input

Grayscale image, size: 32 × 32 pixels.

(MNIST digits were 28 × 28, padded to 32 × 32 for easier convolution).

Layers

1. C1 – Convolutional Layer

*  6 filters of size 5 × 5
*  Output: 28 × 28 × 6 feature maps
*  Captures low-level features (edges, corners).

2. S2 – Subsampling (Pooling) Layer
*  Average pooling with 2 × 2 filter, stride 2
*  output: 14 × 14 × 6
*  Reduces size, keeps important features.

3. C3 – Convolutional Layer

*  16 filters of size 5 × 5
*  Output: 10 × 10 × 16
*  Extracts more complex patterns.

4.  S4 – Subsampling Layer
*  Average pooling with 2 × 2 filter
*  Output: 5 × 5 × 16


5. C5 – Convolutional Layer (Fully Connected in practice)
* 120 filters of size 5 × 5
* Each 5 × 5 filter covers the entire input (5 × 5 × 16).
* Output: 1 × 1 × 120

6.  F6 – Fully Connected Layer
*  84 neurons (inspired by biological neuron connections).

7.  Output Layer
*  10 neurons (for digits 0–9).
*  Uses softmax for classification.

3.  Why was LeNet-5 revulutionery?
  * introduced convolutional+ pooling + fully connected pipeline.
  * Used weight sharing- fewer parameters than fully connected networks.
  * Showed CNN's could learn hierarchical features:
      * Early layers- edges and textures.
      * Later layers-digit parts.
  * First real-world deomsntration of CNN's for document recognition.
  * directly influenced modern CNN architecture like alexnet, which reignited deep learning for computer vision.    

4. Impact on Modern Deep Learning

* AlexNet (2012): Scaled LeNet with more layers and ReLU activation → won ImageNet.

* VGG, ResNet, DenseNet: Built deeper, more efficient architectures, but followed the same convolution → pooling → dense pipeline.

* Inspired applications beyond digits: object detection, face recognition, medical imaging, self-driving cars.

5. Reference (Original Paper)

*  📖 LeCun, Y., Bottou, L., Bengio, Y., & Haffner, P. (1998).
Gradient-based learning applied to document recognition.
Proceedings of the IEEE, 86(11), 2278–2324.

## **Q3** Compare and contrast AlexNet and VGGNet in terms of design principles, number of parameters, and performance. Highlight key innovations and limitations of each.



1. Background

* AlexNet (Krizhevsky et al., 2012)

   * Won the ImageNet Large Scale Visual Recognition Challenge (ILSVRC) 2012.
   * Marked the rebirth of deep learning for computer vision.

* VGGNet (Simonyan & Zisserman, 2014, Oxford Visual Geometry Group)

   * Focused on network depth and simplicity.
   * Achieved top results in ILSVRC 2014 (2nd place).

2. Architectural Design

| Aspect                     | **AlexNet**                           | **VGGNet**                                     |
| -------------------------- | ------------------------------------- | ---------------------------------------------- |
| **Year**                   | 2012                                  | 2014                                           |
| **Input Size**             | 227 × 227 × 3                         | 224 × 224 × 3                                  |
| **Depth**                  | 8 layers (5 conv + 3 fully connected) | 16 or 19 layers (all conv + 3 fully connected) |
| **Convolution Filters**    | 11×11, 5×5, 3×3                       | Only 3×3 (stacked to increase receptive field) |
| **Pooling**                | Overlapping max pooling               | Non-overlapping max pooling                    |
| **Activation**             | ReLU (introduced in CNNs!)            | ReLU                                           |
| **Regularization**         | Dropout (first large-scale use)       | No dropout (relies on depth + augmentation)    |
| **Normalization**          | Local Response Normalization (LRN)    | None                                           |
| **Fully Connected Layers** | 3 (4096–4096–1000)                    | 3 (4096–4096–1000)                             |
| **Parameters**             | ~60 million                           | ~138 million (much larger!)                    |


3. Key Innovations

**AlexNet**

* First to prove CNNs can scale to large datasets (ImageNet, 1.2M images).
* Introduced ReLU activation, making training faster.
* Used dropout to reduce overfitting.
* Trained using GPU acceleration (CUDA) → huge speedup.
* Used data augmentation (cropping, flipping).

**VGGNet**
* Showed depth matters – going deeper improves accuracy.
* Simplified design principle: stack small (3×3) filters → larger effective receptive fields (e.g., two 3×3 convs ≈ one 5×5 conv).
* Very uniform architecture (easy to generalize, unlike AlexNet’s mixed filter sizes).

4. **Limitations**

* AlexNet:

  * Architecture not very systematic (different kernel sizes, overlaps).
  * Still relatively shallow by modern standards.
  * LRN layer later found unnecessary.

* VGGNet:
    * Too large (138M parameters → huge memory + storage).
    * Computationally expensive (training takes weeks without specialized hardware).
    * Replaced later by more efficient deep nets (ResNet, Inception).

##  **Q4**. What is transfer learning in the context of image classification? Explain how it helps in reducing computational costs and improving model performance with limited data.

1. What is Transfer Learning?

Transfer learning in image classification is a technique where a model pre-trained on a large dataset (like ImageNet with 1.2 million images) is reused for a new but related task.

* Instead of training a CNN from scratch, we take a pre-trained network (e.g., VGG, ResNet, Inception) and transfer its learned features (edges, textures, shapes) to solve a new problem (e.g., classifying medical images, plant diseases, or X-rays).

* Usually, the early layers of CNNs learn generic features (edges, corners, textures), while later layers learn task-specific features (faces, digits, cars).

2. How it Reduces Computational Costs

* Training a deep CNN (like VGG16 or ResNet) from scratch requires millions of images and weeks of GPU computation.
* With transfer learning:
   * Most weights are already learned.
   * Only the final few layers (small number of parameters) need training.
   * Saves time, GPU memory, and storage.

👉 Example:

* Training VGG16 from scratch → ~138M parameters.
* Using transfer learning → Only retrain last ~1–2M parameters.

3. How it Improves Performance with Limited Data

* Deep CNNs need huge datasets to generalize well.
* With small datasets (e.g., 1,000 medical images), training from scratch → overfitting.
* Transfer learning uses knowledge from large datasets (ImageNet features like curves, shapes, patterns) → improves accuracy on small datasets.
* Pre-trained features act as good initialization, so the model converges faster and performs better.

## **Q5**. Describe the role of residual connections in ResNet architecture. How do they address the vanishing gradient problem in deep CNNs?  

1. What is ResNet?
* ResNet (Residual Network) was introduced by He et al. (2015) in the paper:
“Deep Residual Learning for Image Recognition” (CVPR 2016).
* It solved the challenge of training very deep CNNs (50, 101, 152 layers) without degradation in performance.

2. The Problem: Vanishing/Degradation
* In very deep networks, gradients get smaller and smaller as they are backpropagated through many layers → vanishing gradient problem.
* This makes it hard to train deep CNNs — adding more layers sometimes makes the network perform worse (degradation).

## **Q6** Implement the LeNet-5 architectures using Tensorflow or PyTorch to classify the MNIST dataset. Report the accuracy and training time.

```python
# lenet5_mnist.py
import time
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
from torchvision import datasets, transforms

# Repro + device
torch.manual_seed(42)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# MNIST: 60k train, 10k test; 28x28 grayscale -> pad to 32x32 for LeNet-5
transform = transforms.Compose([
    transforms.Pad(2),                   # 28->32
    transforms.ToTensor(),               # [0,1]
    transforms.Normalize((0.1307,), (0.3081,))  # standard MNIST norm
])

train_ds = datasets.MNIST(root="./data", train=True, download=True, transform=transform)
test_ds  = datasets.MNIST(root="./data", train=False, download=True, transform=transform)

train_loader = DataLoader(train_ds, batch_size=128, shuffle=True, num_workers=2, pin_memory=True)
test_loader  = DataLoader(test_ds,  batch_size=512, shuffle=False, num_workers=2, pin_memory=True)

# Classic LeNet-5 (slightly modernized with ReLU; original used tanh)
class LeNet5(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.conv1 = nn.Conv2d(1, 6, kernel_size=5, stride=1)     # 32->28
        self.pool1 = nn.AvgPool2d(kernel_size=2, stride=2)        # 28->14
        self.conv2 = nn.Conv2d(6, 16, kernel_size=5, stride=1)    # 14->10
        self.pool2 = nn.AvgPool2d(kernel_size=2, stride=2)        # 10->5
        self.conv3 = nn.Conv2d(16, 120, kernel_size=5, stride=1)  # 5->1
        self.fc1 = nn.Linear(120, 84)
        self.fc2 = nn.Linear(84, num_classes)

    def forward(self, x):
        x = F.relu(self.conv1(x))
        x = self.pool1(x)
        x = F.relu(self.conv2(x))
        x = self.pool2(x)
        x = F.relu(self.conv3(x))      # shape [B,120,1,1]
        x = x.view(x.size(0), -1)      # [B,120]
        x = F.relu(self.fc1(x))        # 84
        x = self.fc2(x)                # 10
        return x

model = LeNet5().to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
criterion = nn.CrossEntropyLoss()

def evaluate(loader):
    model.eval()
    correct, total, loss_sum = 0, 0, 0.0
    with torch.no_grad():
        for x, y in loader:
            x, y = x.to(device), y.to(device)
            logits = model(x)
            loss = criterion(logits, y)
            loss_sum += loss.item() * x.size(0)
            preds = logits.argmax(dim=1)
            correct += (preds == y).sum().item()
            total += x.size(0)
    return loss_sum/total, correct/total

# Train
epochs = 5
start = time.time()
for ep in range(1, epochs+1):
    model.train()
    running = 0.0
    for x, y in train_loader:
        x, y = x.to(device), y.to(device)
        optimizer.zero_grad()
        logits = model(x)
        loss = criterion(logits, y)
        loss.backward()
        optimizer.step()
        running += loss.item() * x.size(0)
    train_loss = running / len(train_loader.dataset)
    val_loss, val_acc = evaluate(test_loader)
    print(f"Epoch {ep}: train_loss={train_loss:.4f}  test_loss={val_loss:.4f}  test_acc={val_acc*100:.2f}%")
end = time.time()

# Final report
test_loss, test_acc = evaluate(test_loader)
print(f"\nFinal Test Accuracy: {test_acc*100:.2f}%")
print(f"Total Training Time: {end-start:.2f} seconds on {device}")
```

## **Q7** Use a pre-trained VGG16 model (via transfer learning) on a small custom dataset (e.g., flowers or animals). Replace the top layers and fine-tune the model. Include your code and result discussion.

```python
# vgg16_flowers102_transfer.py
import time, copy
import torch
import torch.nn as nn
from torch.optim import lr_scheduler
from torchvision import models, transforms
from torchvision.datasets import Flowers102
from torch.utils.data import DataLoader

# Repro + device
torch.manual_seed(42)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Flowers102 splits are ("train","val","test"); auto-download is supported
# Input pipeline: resize->tensor->ImageNet normalize for VGG16
input_size = 224
normalize_mean = [0.485, 0.456, 0.406]
normalize_std  = [0.229, 0.224, 0.225]

train_tf = transforms.Compose([
    transforms.Resize((input_size, input_size)),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(0.1,0.1,0.1,0.05),
    transforms.ToTensor(),
    transforms.Normalize(normalize_mean, normalize_std),
])
val_tf = transforms.Compose([
    transforms.Resize((input_size, input_size)),
    transforms.ToTensor(),
    transforms.Normalize(normalize_mean, normalize_std),
])

root = "./data"

train_ds = Flowers102(root=root, split="train", transform=train_tf, download=True)
val_ds   = Flowers102(root=root, split="val",   transform=val_tf,   download=True)
test_ds  = Flowers102(root=root, split="test",  transform=val_tf,   download=True)

train_loader = DataLoader(train_ds, batch_size=32, shuffle=True,  num_workers=4, pin_memory=True)
val_loader   = DataLoader(val_ds,   batch_size=64, shuffle=False, num_workers=4, pin_memory=True)
test_loader  = DataLoader(test_ds,  batch_size=64, shuffle=False, num_workers=4, pin_memory=True)

num_classes = 102  # Flowers102 has 102 categories
dataloaders = {"train": train_loader, "val": val_loader}
sizes = {"train": len(train_ds), "val": len(val_ds)}

# Load pretrained VGG16 and replace classifier head
model = models.vgg16(weights=models.VGG16_Weights.IMAGENET1K_V1).to(device)
in_features = model.classifier[6].in_features
model.classifier[6] = nn.Linear(in_features, num_classes).to(device)

# Stage 1: freeze features, train head
for p in model.features.parameters():
    p.requires_grad = False

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(filter(lambda p: p.requires_grad, model.parameters()), lr=1e-3)
scheduler = lr_scheduler.StepLR(optimizer, step_size=5, gamma=0.1)

def run_epoch(model, loader, train=True):
    if train:
        model.train()
    else:
        model.eval()
    running_loss, running_corrects = 0.0, 0
    with torch.set_grad_enabled(train):
        for x, y in loader:
            x, y = x.to(device), y.to(device)
            if train:
                optimizer.zero_grad()
            logits = model(x)
            loss = criterion(logits, y)
            if train:
                loss.backward()
                optimizer.step()
            running_loss += loss.item() * x.size(0)
            running_corrects += (logits.argmax(1) == y).sum().item()
    return running_loss / len(loader.dataset), running_corrects / len(loader.dataset)

def train_model(model, epochs):
    since = time.time()
    best_wts = copy.deepcopy(model.state_dict())
    best_acc = 0.0
    for ep in range(1, epochs+1):
        tr_loss, tr_acc = run_epoch(model, dataloaders["train"], train=True)
        scheduler.step()
        va_loss, va_acc = run_epoch(model, dataloaders["val"], train=False)
        print(f"Epoch {ep}: train_loss={tr_loss:.4f} acc={tr_acc*100:.2f}% | val_loss={va_loss:.4f} acc={va_acc*100:.2f}%")
        if va_acc > best_acc:
            best_acc = va_acc
            best_wts = copy.deepcopy(model.state_dict())
    total = time.time() - since
    model.load_state_dict(best_wts)
    return model, best_acc, total

print("Stage 1: train classifier head only...")
model, head_best_acc, head_time = train_model(model, epochs=5)

# Stage 2: unfreeze last conv block (conv5) and fine-tune with a lower LR
# VGG16 features: blocks roughly end at indices [4,9,16,23,30]; conv5 is ~24:31
for p in model.features[24:].parameters():
    p.requires_grad = True

optimizer = torch.optim.Adam(filter(lambda p: p.requires_grad, model.parameters()), lr=1e-4)
scheduler = lr_scheduler.StepLR(optimizer, step_size=5, gamma=0.1)

print("Stage 2: fine-tune last conv block...")
model, ft_best_acc, ft_time = train_model(model, epochs=5)

# Final test evaluation
def evaluate_test(model, loader):
    model.eval()
    correct, total = 0, 0
    with torch.no_grad():
        for x, y in loader:
            x, y = x.to(device), y.to(device)
            logits = model(x)
            correct += (logits.argmax(1) == y).sum().item()
            total += x.size(0)
    return correct / total

test_acc = evaluate_test(model, test_loader)
print(f"\nSummary:")
print(f"Head best val acc: {head_best_acc*100:.2f}% in {head_time:.1f}s")
print(f"FT   best val acc: {ft_best_acc*100:.2f}% in {ft_time:.1f}s")
print(f"Test accuracy: {test_acc*100:.2f}%")
torch.save(model.state_dict(), "vgg16_flowers102_finetuned.pth")
print(f"Saved weights to vgg16_flowers102_finetuned.pth")
```

## **Q8** Write a program to visualize the filters and feature maps of the first convolutional layer of AlexNet on an example input image.

```python
# alexnet_filters_featuremaps.py
import torch
import torchvision.transforms as T
from torchvision import models
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np

# 1) Load pretrained AlexNet
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
alex = models.alexnet(weights=models.AlexNet_Weights.IMAGENET1K_V1).to(device).eval()  # conv1->ReLU->pool follows [web:198]

# 2) Image preprocess: resize/crop to 224 and normalize to ImageNet stats [web:197][web:199]
preprocess = T.Compose([
    T.Resize(256),
    T.CenterCrop(224),
    T.ToTensor(),
    T.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])

# Provide a local image path
img_path = "example.jpg"  # replace with any RGB image
img = Image.open(img_path).convert("RGB")
x = preprocess(img).unsqueeze(0).to(device)  # [1,3,224,224]

# 3) Extract first conv layer weights (filters)
conv1 = alex.features[0]  # nn.Conv2d(3,64,kernel_size=11,stride=4,padding=2) [web:198][web:193]
w = conv1.weight.detach().cpu().numpy()      # [64,3,11,11]

# Helper to grid-plot a list/array of 2D maps
def save_grid(maps, grid_rows, grid_cols, fname, cmap="gray"):
    plt.figure(figsize=(grid_cols*2, grid_rows*2))
    for i in range(grid_rows*grid_cols):
        if i >= len(maps): break
        ax = plt.subplot(grid_rows, grid_cols, i+1)
        ax.imshow(maps[i], cmap=cmap)
        ax.axis("off")
    plt.tight_layout()
    plt.savefig(fname, dpi=200, bbox_inches="tight")
    plt.close()

# 4) Visualize filters: convert each 3x11x11 filter to RGB for display by min-max per filter
filters_rgb = []
for i in range(w.shape[0]):                  # 64 filters
    f = w[i]                                 # [3,11,11]
    f_min, f_max = f.min(), f.max()
    f_norm = (f - f_min) / (f_max - f_min + 1e-8)
    f_img = np.transpose(f_norm, (1,2,0))    # [11,11,3]
    filters_rgb.append(f_img)
save_grid(filters_rgb, grid_rows=8, grid_cols=8, fname="filters_conv1.png", cmap=None)  # 8x8 grid

# 5) Register a forward hook to grab feature maps after conv1
feat_maps = {}
def hook_fn(module, inp, out):
    feat_maps["conv1_out"] = out.detach().cpu()  # [1,64,H,W]
hook_handle = conv1.register_forward_hook(hook_fn)

# Forward pass
with torch.no_grad():
    _ = alex(x)
hook_handle.remove()

fm = feat_maps["conv1_out"][0].numpy()  # [64,H,W]
# Normalize each channel for display
fm_imgs = []
for i in range(fm.shape[0]):
    m = fm[i]
    m_min, m_max = m.min(), m.max()
    m_norm = (m - m_min) / (m_max - m_min + 1e-8)
    fm_imgs.append(m_norm)
save_grid(fm_imgs, grid_rows=8, grid_cols=8, fname="featuremaps_conv1.png", cmap="viridis")

print("Saved filters_conv1.png and featuremaps_conv1.png")
```

## **Q9**  Train a GoogLeNet (Inception v1) or its variant using a standard dataset like CIFAR-10. Plot the training and validation accuracy over epochs and analyze overfitting or underfitting.

```python
# googlenet_cifar10.py
import time, copy
import torch, torch.nn as nn
from torch.utils.data import DataLoader
from torchvision import datasets, transforms, models
import matplotlib.pyplot as plt

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
torch.manual_seed(42)

# 1) Data: CIFAR-10 (auto-download). Normalize to CIFAR convention. [web:201][web:209]
tf_train = transforms.Compose([
    transforms.RandomCrop(32, padding=4),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize((0.4914,0.4822,0.4465),(0.2470,0.2435,0.2616)),
])
tf_test = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.4914,0.4822,0.4465),(0.2470,0.2435,0.2616)),
])

train_ds = datasets.CIFAR10("./data", train=True,  download=True, transform=tf_train)
val_ds   = datasets.CIFAR10("./data", train=False, download=True, transform=tf_test)

train_loader = DataLoader(train_ds, batch_size=128, shuffle=True,  num_workers=2, pin_memory=True)
val_loader   = DataLoader(val_ds,   batch_size=256, shuffle=False, num_workers=2, pin_memory=True)
sizes = {"train": len(train_ds), "val": len(val_ds)}

# 2) Model: GoogLeNet for 10 classes with aux logits (helps training). [web:210]
model = models.googlenet(num_classes=10, aux_logits=True).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.05, momentum=0.9, weight_decay=5e-4)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=20)

def step(loader, train=True):
    model.train(train)
    total_loss, correct = 0.0, 0
    for x,y in loader:
        x,y = x.to(device), y.to(device)
        if train: optimizer.zero_grad()
        out = model(x)
        # With aux_logits=True, model outputs a namedtuple(logits, aux2, aux1) in train mode. [web:210]
        if train and isinstance(out, tuple):
            logits, aux2, aux1 = out
            loss = criterion(logits, y) + 0.3*criterion(aux1, y) + 0.3*criterion(aux2, y)
        else:
            logits = out.logits if hasattr(out, "logits") else out
            loss = criterion(logits, y)
        if train:
            loss.backward()
            optimizer.step()
        total_loss += loss.item()*x.size(0)
        correct += (logits.argmax(1)==y).sum().item()
    return total_loss/len(loader.dataset), correct/len(loader.dataset)

# 3) Train loop
epochs = 20
hist = {"train_acc":[], "val_acc":[]}
best_wts, best_val = copy.deepcopy(model.state_dict()), 0.0
start = time.time()
for ep in range(1, epochs+1):
    tr_loss, tr_acc = step(train_loader, train=True)
    va_loss, va_acc = step(val_loader,   train=False)
    scheduler.step()
    hist["train_acc"].append(tr_acc); hist["val_acc"].append(va_acc)
    if va_acc>best_val: best_val=va_acc; best_wts=copy.deepcopy(model.state_dict())
    print(f"Epoch {ep:02d}: train_acc={tr_acc*100:.2f}% val_acc={va_acc*100:.2f}%")

total_time = time.time()-start
model.load_state_dict(best_wts)
print(f"\nBest Val Acc: {best_val*100:.2f}% | Time: {total_time/60:.1f} min on {device}")

# 4) Plot accuracy curves
plt.figure(figsize=(6,4))
plt.plot(range(1,epochs+1), [a*100 for a in hist['train_acc']], label="Train")
plt.plot(range(1,epochs+1), [a*100 for a in hist['val_acc']],   label="Val")
plt.xlabel("Epoch"); plt.ylabel("Accuracy (%)"); plt.title("GoogLeNet on CIFAR-10")
plt.legend(); plt.tight_layout(); plt.savefig("googlenet_cifar10_acc.png", dpi=180)
print("Saved plot: googlenet_cifar10_acc.png")
```
